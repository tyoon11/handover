# KHS gold(`인계요약지`) 채움 확인
다중헤더(3행) 파일. c11 `인계요약지`가 교수님이 새로 쓴 진짜 gold인지(채워졌는지) 확인.
셀 하나 실행 후 전체 출력 공유.

```python
import pandas as pd

PATH = "/home/coder/workspace/data/handover/data/gold_sampled/인계요약지_gold_sampled_251002_KHS.xlsx"
HEADER_ROWS = 3  # row0~2 = 다중헤더, 데이터는 row3~
# 위치 기반 컬럼: 0=idx,1=환자ID,2=sid,3=진료과,4=수술진료과,5=퇴실,6=마취기록,
#               7=마취전요약,8=premed,9=LLM참고1,10=LLM참고2,11=인계요약지(gold)
C_IDX, C_SID, C_LLM1, C_LLM2, C_GOLD = 0, 2, 9, 10, 11

raw = pd.read_excel(PATH, sheet_name="데이터", header=None)
data = raw.iloc[HEADER_ROWS:].reset_index(drop=True)
print("전체 raw shape:", raw.shape, "| 데이터 행수:", len(data))

def filled(col):
    s = data.iloc[:, col]
    return s.apply(lambda x: pd.notna(x) and str(x).strip() not in ("", "nan", "-"))

for name, col in [("LLM참고1(c9)", C_LLM1), ("LLM참고2(c10)", C_LLM2), ("인계요약지 gold(c11)", C_GOLD)]:
    print(f"  {name}: 채워진 행 {int(filled(col).sum())}/{len(data)}")

print("\n── c11 인계요약지(gold) 전체 샘플 ──")
g = filled(C_GOLD)
for i in range(len(data)):
    if g.iloc[i]:
        sid = data.iloc[i, C_SID]
        print(f"\n[row {i}] sid={sid}\n  GOLD: {str(data.iloc[i, C_GOLD])[:300]}")
```
